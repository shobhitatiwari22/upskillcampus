```python
# 1. Define the columns from the empty early years that we want to drop
columns_to_drop = [
    ' 3-1993', ' 3-1994', ' 3-1995', ' 3-1996', ' 3-1997', 
    ' 3-1998', ' 3-1999', ' 3-2000', ' 3-2001', ' 3-2002', ' 3-2003'
]

# 2. Drop those highly empty columns
produce_cleaned = produce.drop(columns=columns_to_drop)

# 3. For the remaining good years (2004-2014), drop rows that still have missing data
produce_cleaned = produce_cleaned.dropna()

# 4. Verify the final clean structure
print("--- Cleaned Produce Dataset Info ---")
print(produce_cleaned.info())
```

```python
# 1. First, strip all leading spaces from the original produce columns permanently
produce.columns = [col.strip() for col in produce.columns]

# 2. Reshape the dataset from wide format to long format
produce_long = pd.melt(
    produce, 
    id_vars=['Particulars', 'Frequency', 'Unit'], 
    var_name='Year', 
    value_name='Production'
)

# 3. Clean the 'Year' column to remove the '3-' prefix so it is just a pure number (e.g., 1993)
produce_long['Year'] = produce_long['Year'].str.replace('3-', '', regex=False)
produce_long['Year'] = pd.to_numeric(produce_long['Year'])

# 4. Drop ONLY the specific rows where production is missing, keeping everything else!
produce_long = produce_long.dropna(subset=['Production'])

# 5. Let's inspect our beautiful, cleaned long-format data
print(f"Total remaining high-quality records: {len(produce_long)}")
produce_long.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
```

```python
features= ['Crop', 'State', 'Cost_of_Cultivation_C2', 'Cost_of_Production_C2']
X= datafile1[features]
y= datafile1['Yield']
```

```python
X_encoded= pd.get_dummies(X, columns=['Crop', 'State'], drop_first=True)
```

```python
X_train, X_test, y_train, y_test= train_test_split(X_encoded, y, test_size=0.2, random_state=42)
model= RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
accuracy = model.score(X_test, y_test)
print(f"🎯 Model Prediction Accuracy: {accuracy * 100:.1f}%\n")
```

```python
# 6. Show a simple comparison table of Actual vs Predicted
predictions = model.predict(X_test)
comparison_df = pd.DataFrame({'Actual Yield': y_test, 'Predicted Yield': predictions.round(1)})
print("--- Sample Predictions Check ---")
print(comparison_df.head())
```

## Actual Vs Predicted Chart

```python
plot_df = comparison_df.reset_index(drop=True)

plt.figure(figsize=(10, 5))

plt.plot(plot_df['Actual Yield'], label='Actual Yield', color='blue', marker='o', linewidth=2)

plt.plot(plot_df['Predicted Yield'], label='Predicted Yield', color='orange', marker='x', linestyle='--', linewidth=2)

plt.title('Actual vs Predicted Crop Yield (96.6% Accuracy)', fontsize=14, fontweight='bold')
plt.xlabel('Test Data Samples', fontsize=12)
plt.ylabel('Yield (Quintals per Hectare)', fontsize=12)

plt.legend(fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import pickle
import numpy as np

st.set_page_config(page_title="Indian Crop Yield Predictor", page_icon="🌾", layout="centered")

st.title("🌾 Indian Crop Yield Prediction App")
st.write("This interactive application uses a **Random Forest Regressor (96.6% Accuracy)** to predict agricultural crop yields based on input metrics.")

@st.cache_resource
def load_assets():
    with open("crop_model.pk1", "rb") as model_file:
        model = pickle.load(model_file)
    with open("model_columns.pk1", "rb") as columns_file:
        model_columns = pickle.load(columns_file)
    return model, model_columns

try:
    model, model_columns = load_assets()
except FileNotFoundError:
    st.error("⚠️ Model files not found! Please check that 'crop_model.pk1' and 'model_columns.pk1' are in the same folder.")
    st.stop()

crops = ['ARHAR', 'COTTON', 'GRAM', 'GROUNDNUT', 'MAIZE', 'MOONG', 'PADDY', 'RAPESEED AND MUSTARD', 'SUGARCANE', 'WHEAT']
states = ['Uttar Pradesh', 'Karnataka', 'Gujarat', 'Andhra Pradesh', 'Maharashtra', 'Punjab', 'Haryana', 'Rajasthan', 'Madhya Pradesh', 'Tamil Nadu', 'Bihar', 'Orissa', 'West Bengal']


st.subheader("📋 Enter Prediction Parameters")

col1, col2 = st.columns(2)
with col1:
    selected_crop = st.selectbox("Select Crop Type:", sorted(crops))
    cost_c2 = st.number_input("Cost of Cultivation C2 (Rs./Hectare):", min_value=0.0, value=25000.0, step=500.0)

with col2:
    selected_state = st.selectbox("Select Indian State:", sorted(states))
    cost_prod_c2 = st.number_input("Cost of Production C2 (Rs./Quintal):", min_value=0.0, value=2000.0, step=50.0)

if st.button("🔮 Predict Expected Crop Yield"):
    # Generate a baseline input row filled with zeros matching your original training columns
    input_data = pd.DataFrame(0, index=[0], columns=model_columns)
    
    # Populate numerical variables
    if 'Cost_of_Cultivation_C2' in input_data.columns:
        input_data['Cost_of_Cultivation_C2'] = cost_c2
    if 'Cost_of_Production_C2' in input_data.columns:
        input_data['Cost_of_Production_C2'] = cost_prod_c2
        
    # Populate One-Hot Encoded variables (matching dummy structures drop_first dynamic)
    crop_col = f"Crop_{selected_crop}"
    state_col = f"State_{selected_state}"
    
    if crop_col in input_data.columns:
        input_data[crop_col] = 1
    if state_col in input_data.columns:
        input_data[state_col] = 1
        
    # Execute model prediction
    prediction = model.predict(input_data)[0]
    
    # Display the result to the user
    st.success(f"### 🌱 Estimated Crop Yield: **{prediction:.2f} Quintals per Hectare**")
```

### What Exactly Are We Predicting & Why? 
#### 1. What is the Target Variable?
We are predicting **Crop Yield** (measured in Quintals per Hectare). 
* **Yield** tells us the **efficiency and productivity** of a farm. It answers the question: *"How much physical crop can a farmer successfully harvest from a single piece of land?"*

#### 2. Why do we need "Model Predicted Yield" instead of just using the "Historical Average"?
* **The Real Historical Average** is a fixed number looking backward. It only tells you what a typical crop produced in the past across the entire country, regardless of conditions. 
* **The Model Predicted Yield** is a dynamic, intelligent future forecast. It looks at the **exact crop**, the **exact state**, and the **specific budget constraints (Cultivation & Production Costs)** that you enter right now. 

#### 3. The Real-World "Production" Impact
Our machine learning model acts as a **smart farm simulator**. By changing the inputs, stakeholders can instantly see the results:
* **For Farmers:** It helps them see how cutting or increasing their cultivation budget will directly impact their final harvest volume.
* **For Banks & Supply Managers:** It allows them to forecast food supply levels and calculate financial risks before the seeds are even planted.
